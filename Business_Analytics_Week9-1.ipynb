## Decision Tree

### 0. Data

```python
url = "http://byungwan.com/class/House_Prices.xls"
```

```python
import pandas as pd

house = pd.read_excel(url)
house.head()
```

### 1. Converting Numerical to Categorical

```python
house["Price"].median()
```

```python
house["House_Class"] = pd.qcut(house["Price"], 2, labels=[0,1])
```

```python
house.head()
```

### 2. Converting Categorical to Numerical

```python
house = pd.get_dummies(house, columns=["Brick", "Neighborhood"], drop_first=True, dtype=int)
```

```python
house.head()
```

### 3. Define Y and X

```python
house.columns.get_loc("House_Class")
```

```python
y = house.iloc[:, 6]
x = house.iloc[:, [2,3,4,5,7,8,9]]
```

```python
from sklearn.model_selection import train_test_split

x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.3, random_state=0) 

# test size = 0.3(30%) for testing, remaining(70%) for training
# no matter what value is used for random_state, for as long as we use the samve value, the result will be consistent
```

### 4. Decision Tree

```python
from sklearn.tree import DecisionTreeClassifier

dtc = DecisionTreeClassifier()
```

```python
dt_model = dtc.fit(x_train, y_train)
```

```python
y_pred = dt_model.predict(x_test)
```

```python
from sklearn import metrics

metrics.accuracy_score(y_test, y_pred)

# accuracy = (TP + TN) / Total
```
