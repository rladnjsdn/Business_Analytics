## Neural Network

## 0. Data

```python
import pandas as pd

url = "http://byungwan.com/class/House_Prices2.xlsx"
house = pd.read_excel(url)

house.head()
```

```python
house.columns.get_loc("House_Class")
```

```python
y = house.iloc[:, 6]
x = house.iloc[:, [2,3,4,5,7,8,9]]
```

```python
from sklearn.model_selection import train_test_split

x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.3, random_state=0)

x.describe()
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
scaler_model = scaler.fit(x_train)
scaled_x_train = scaler_model.transform(x_train)

x_train.describe()
```

```python
pd.DataFrame(scaled_x_train, columns=x.columns).describe()
```

## 2. Artificial Neural Network (ANN) / Multi-Layer Perceptron (MLP)

```python
from sklearn.neural_network import MLPClassifier

mlp = MLPClassifier(activation="relu", hidden_layer_sizes=(1), max_iter=3000, random_state=0) # activation = sigmoid / tanh / relu
```

```python
mlp_model = mlp.fit(scaled_x_train, y_train)
```

```python
scaled_x_test = scaler_model.transform(x_test)
```

```python
y_pred = mlp_model.predict(scaled_x_test)
```

```python
from sklearn import metrics

metrics.accuracy_score(y_test, y_pred)
```

#### # of nodes in a hidden layer

optimal no. of nodes = avg(input_layer + output_layer) = 7(x) + 1(y) / 2 = 4

```python
from sklearn.model_selection import cross_val_score
import numpy as np

scaled_x = scaler.fit_transform(x)
size = range(1, 8)

for s in size:
    mlp_n = MLPClassifier(activation="relu", hidden_layer_sizes=(s), max_iter=3000, random_state=0)
    scores = cross_val_score(mlp_n, scaled_x, y, cv=5)
    mean = np.mean(scores)
    print(str(s) + ": " + str(mean))
```

### 4. Multiple Hidden Layers

```python
mlp = MLPClassifier(activation="relu", hidden_layer_sizes=(1,1,1), max_iter=3000, random_state=0) # 3 hidden layers and 1 node in each hidden layer
mlp_model = mlp.fit(scaled_x_train, y_train)
y_pred = mlp_model.predict(scaled_x_test)
metrics.accuracy_score(y_test, y_pred)
```

#### # of nodes and # of hidden layers

```python
import itertools

a = b = c = range(1,4)
abcs = list(itertools.product(a,b,c))
```

```python
abcs
```

```python
for abc in abcs:
    mlp_n = MLPClassifier(activation="relu", hidden_layer_sizes=(abc), max_iter=3000, random_state=0)
    score = cross_val_score(mlp_n, scaled_x, y, cv=5)
    mean = np.mean(score)
    print(str(abc) + ": " + str(mean))
```
