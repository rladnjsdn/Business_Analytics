```python
import pandas as pd

url = "http://byungwan.com/class/House_Prices.xls"
house = pd.read_excel(url)

house.head()
```

```python
house["House_Class"] = pd.qcut(house["Price"], 2, labels=[0,1])
```

```python
house.head(2)
```

```python
house = pd.get_dummies(house, columns=["Brick", "Neighborhood"], drop_first=True, dtype=int)
```

```python
house.head(2)
```

```python
y = house.iloc[:, 6]
x = house.iloc[:, [2,3,4,5,7,8,9,]]
```

```python
from sklearn.model_selection import train_test_split

x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.3, random_state=100)
```

```python
from sklearn.tree import DecisionTreeClassifier

dtc = DecisionTreeClassifier(criterion="gini", max_depth=3)
```

```python
dt_model = dtc.fit(x_train, y_train)
y_pred = dt_model.predict(x_test)
```

```python
from sklearn import metrics

metrics.accuracy_score(y_test, y_pred)
```

```python
from sklearn.model_selection import cross_val_score
```

```python
scores = cross_val_score(dtc, x, y, cv=5)
```

```python
import numpy as np

np.mean(scores)
```

```python
url2 = "http://byungwan.com/class/Coupon_Sample.xlsx"

coupon = pd.read_excel(url2)

coupon.head()
```

```python
coupon["Discount"].value_counts()
```

## Upsampling

```python
coupon0 = coupon[coupon["Discount"] == 0]
coupon1 = coupon[coupon["Discount"] == 1]

coupon0["Discount"].value_counts()
```

```python
coupon1["Discount"].value_counts()
```

```python
from sklearn.utils import resample
```

```python
coupon1_up = resample(coupon1, replace=True, n_samples=377, random_state=0)
```

```python
coupon1_up["Discount"].value_counts() 
```

```python
coupon_up = pd.concat([coupon0, coupon1_up], axis=0)
```

```python
coupon_up["Discount"].value_counts()
```
