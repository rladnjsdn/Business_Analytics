```python
import pandas as pd
```

```python
url = "http://byungwan.com/class/Coupon_Sample.xlsx"
```

```python
coupon = pd.read_excel(url)
```

```python
coupon.head(10)
```

```python
from matplotlib import pyplot as plt
```

## 1. Box Chart

```python
plt.figure(figsize=(2,2))
plt.boxplot(coupon["Sales"], tick_labels=["Sales"])
plt.show()
```

```python
plt.figure(figsize=(2,2))
plt.boxplot(coupon["Sales"], tick_labels=["Sales"], notch=True, vert=False)  # Notch = Median Value inside the box , Vert = Vertical / Horizontal Box Chart
plt.show()
```

```python
coupon["Sales"].describe()
```

## 2. Histogram

```python
plt.figure(figsize=(2,2))
plt.hist(coupon["Sales"], bins = "auto", facecolor = "lightblue")
plt.xlabel("Sales", fontsize=12)
plt.ylabel("Frequency", fontsize=12)
plt.show()
```

## 3. Scatter Plot

```python
plt.figure(figsize=(3,2))
plt.scatter(x = coupon["Sales"], y = coupon["Gender"], s = coupon["Visits"]*10, alpha=0.5) # s = size , alpha = transparency
plt.yticks([0,1], ["Female", "Male"]) # 0 = Female , 1 = Male
plt.xlabel("Sales", fontsize = 12)
plt.ylabel("Gender", fontsize = 12)
plt.margins(0.1, 0.3)
plt.grid(color="lightgray")
plt.show()
```

## 4. Summary Statistics

```python
coupon.describe()
```

## 5. Correlation

```python
corrs = coupon.corr()
```

```python
import seaborn as sns
```

```python
plt.figure(figsize=(10,10))
sns.heatmap(corrs, vmin=-1, vmax=1, center=0, cmap="coolwarm", annot=True) # -1 = negatively correlated , 0 = no correlation, 1 = positively correlated
plt.show()
```
