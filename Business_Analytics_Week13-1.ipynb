## Support Vector Machine

### 0. Data

```python
import pandas as pd

url = "http://byungwan.com/class/House_Prices2.xlsx"
house = pd.read_excel(url)
house.head()
```

```python
y = house.iloc[:, 6]
x = house.iloc[:, [2,3,4,5,7,8,9]]
```

```python
from sklearn.model_selection import train_test_split

x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.3, random_state=0)
```

### 1. Linear Support Vector Machine (SVM)

```python
from sklearn.svm import SVC
from sklearn import metrics

lsvc = SVC(kernel = "linear")
lsvc_model = lsvc.fit(x_train, y_train)
y_pred = lsvc_model.predict(x_test)

metrics.accuracy_score(y_test, y_pred)
```

### 2. Non-linear SVC

```python
nlsvc = SVC(kernel = "rbf") #non-linear kernels = poly / sigmoid / rbf (gaussian)
nlsvc_model = nlsvc.fit(x_train, y_train)
y_pred = nlsvc_model.predict(x_test)

metrics.accuracy_score(y_test, y_pred)
```

### 2.1 Parameter Tuning

```python
nlsvc = SVC(kernel = "rbf", C = 0.01, gamma = 0.01) # gamma level = how much curvature for the line
nlsvc_model = nlsvc.fit(x_train, y_train)
y_pred = nlsvc_model.predict(x_test)

metrics.accuracy_score(y_test, y_pred)
```

### 2.2 Grid Search

```python
from sklearn.model_selection import cross_val_score
from sklearn.model_selection import GridSearchCV
```

```python
params = {"C": [0.001, 0.01, 0.1, 1, 10, 100, 1000],
          "gamma": [0.001, 0.01, 0.1, 1, 10, 100, 1000]}
grid = GridSearchCV(SVC(kernel = "rbf"), params, cv=5)
grid_model = grid.fit(x, y)
grid_model.best_params_
```

```python
grid_model.best_score_
```

```python
import pandas as pd

url = "http://byungwan.com/class/Coupon_Sample.xlsx"
coupon = pd.read_excel(url)
coupon.head()
```

```python
y = coupon.iloc[:, 4]
x = coupon.iloc[:, [1,2,3,5,6,7,8,9,10]]
```

```python
from sklearn.svm import SVC
from sklearn.model_selection import cross_val_score
from sklearn.model_selection import GridSearchCV

params = {"kernel": ["rbf", "sigmoid"],
          "C": [0.001, 0.01, 0.1, 1, 10, 100, 1000],
          "gamma": [0.001, 0.01, 0.1, 1, 10, 100, 1000]}
grid = GridSearchCV(SVC(), params, cv=5)
grid_model = grid.fit(x, y)
grid_model.best_params_
```

```python
float(grid_model.best_score_)
```
