## Data Management

```python
import pandas as pd
```

```python
house = pd.read_excel("http://byungwan.com/class/House_Prices.xls")
```

```python
house.head()
```

```python
house2 = pd.get_dummies(house, columns=["Brick", "Neighborhood"],drop_first=True, dtype=int)

#pd.get_dummies = Categorical -> Numerical
```

## Converting numerical to categorical

```python
house2.describe()
```

```python
price_category1 = pd.cut(house2["Price"], bins=[0, 130000, 300000], labels=["Low", "High"]) #creates column from 0~130,000 and 130,000~300,000

# pd.cut = numerical -> categorical
```

```python
house2["Price_Category1"] = price_category1
```

```python
house2.head(10)
```

```python
price_category2 = pd.cut(house2["Price"], bins=[0, 100000, 150000, 300000], labels=["Low","Middle","High"])
```

```python
house2["Price_Category2"] = price_category2
```

```python
house2.head(10)
```

```python
price_category3 = pd.cut(house2["Price"], bins = 2, labels=["Low", "High"])
```

```python
house2["Price_Category3"] = price_category3
```

```python
house2.head(10)
```

```python
pd.cut(house2["Price"], bins=2)

# Minimum Value = 69,,100
# Maximum Value = 211,200 

# (Max + Min) / 2 = Threshold for the bin categories
```

```python
house2["Price_Category3"].value_counts()
```

```python
price_category4 = pd.qcut(house2["Price"], 2, labels = ["Low", "High"])

# qcut = quantile cut (based on the value of %)
```

```python
house2["Price_Category4"] = price_category4
```

```python
house2.head(10)
```

```python
house2["Price_Category4"].value_counts()
```

## Data Visualization

```python
url = "http://byungwan.com/class/Coupon_Sample.xlsx"
```

```python
coupon = pd.read_excel(url)
```

```python
coupon.head()
```

## 1. Bar Chart

```python
from matplotlib import pyplot as plt
```

```python
coupon["Gender"].value_counts()
```

```python
gender_freq = coupon["Gender"].value_counts().values
```

```python
gender_freq
```

```python
gender_x = ["Female", "Male"]
```

```python
plt.figure(figsize=(2,2))
plt.bar(gender_x, gender_freq, color="lightblue")
plt.xlabel("Gender", fontsize=12)
plt.ylabel("Count (persons)", fontsize=12)
plt.show()
```

## 2. Pie Chart

```python
plt.figure(figsize=(2,2))
plt.pie(gender_freq, labels=gender_x, autopct="%.2f%%") #autopct = auto percentage, .2f = up to 2 decimals points 
plt.show()
```

```python
plt.figure(figsize=(2,2))
plt.pie(gender_freq, labels = gender_x, autopct="%.1f%%", colors = ["lightblue", "coral"], explode=[0.1, 0])
plt.show()
```
