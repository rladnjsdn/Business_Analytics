## Neural Networks

### 0. Data

```python
import pandas as pd

url = "http://byungwan.com/class/Coupon_Sample.xlsx"

coupon = pd.read_excel(url)
coupon.head()
```

```python
coupon["Discount"].value_counts()
```

```python
from sklearn.utils import resample

coupon0 = coupon[coupon["Discount"] == 0]
coupon1 = coupon[coupon["Discount"] == 1]

coupon0_down = resample(coupon0, replace=False, n_samples=122, random_state=0)
coupon_down = pd.concat([coupon0_down, coupon1], axis=0)

coupon_down["Discount"].value_counts()
```

```python
coupon_down.head()
```

```python
y = coupon_down.iloc[:, 4]
x = coupon_down.iloc[:, [1,2,3,5,6,7,8,9,10]]
```

```python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
```

```python
scaler_model = scaler.fit(x)
scaled_x = scaler_model.transform(x)
```

### 1. ANN

```python
from sklearn.neural_network import MLPClassifier
from sklearn.model_selection import cross_val_score
import numpy as np

mean = []
sizes = range(3, 7)

for s in sizes:
    mlp_n = MLPClassifier(activation="relu", hidden_layer_sizes=(s), max_iter=3000, random_state=0)
    scores = cross_val_score(mlp_n, scaled_x, y, cv=5)
    m = np.mean(scores)
    mean.append(m)
```

```python
mean
```

```python
mlp = MLPClassifier(activation="relu", hidden_layer_sizes=(5), max_iter=3000, random_state=0)
mlp_model = mlp.fit(scaled_x, y)
```

### 2. Evaluation

```python
url = "http://byungwan.com/class/Coupon_Test.xlsx"
coupon_test = pd.read_excel(url)

coupon_test.head()
```

```python
y_test = coupon_test.iloc[:, 4]
x_test = coupon_test.iloc[:, [1,2,3,5,6,7,8,9,10]]

scaled_x_test = scaler_model.transform(x_test)
y_pred = mlp_model.predict(scaled_x_test)
print(y_pred)
```

```python
from sklearn import metrics
metrics.accuracy_score(y_test, y_pred)
```
